## Step 3: データ読み込みとモデル学習

選択したデータセットでモデルを学習します。

```python
# データの読み込み
# chestmnistの場合は2値分類に変換
if data_flag == 'chestmnist':
    (x_train, y_train), (x_test, y_test), info = exercise_logic.load_and_preprocess(
        data_flag, 
        binary_classification=True
    )
    num_classes = 1
else:
    (x_train, y_train), (x_test, y_test), info = exercise_logic.load_and_preprocess(data_flag)
    num_classes = len(info['label'])

print(f"データセット: {data_flag}")
print(f"訓練データ: {x_train.shape}")
print(f"ラベル一覧: {info['label']}")
```

```python
# モデルの構築と学習
model = exercise_logic.build_model(
    input_shape=(28, 28, 3), 
    num_classes=num_classes,
    model_type='simple'
)

print("モデルを学習中...（約2-3分）")
history = model.fit(
    x_train, y_train, 
    epochs=5, 
    validation_split=0.1, 
    batch_size=128,
    verbose=1
)
print("\n✓ モデル学習が完了しました。")
```

## 考察課題の回答例

以下は考察課題に対する回答の一例です。これが唯一の正解ではなく、議論の出発点として活用してください。

### 1. Webアプリ公開のメリット

AIをWebアプリとして提供する利点：

- 利用者側の環境構築負荷を小さくできる（ブラウザで利用可能）
- 入力→結果までが短く、フィードバックが得やすい
- URL共有で試用・評価が容易
- サーバー側で更新でき、改善を配布しやすい（ただし変更管理と性能監視が重要）

### 2. 医療AIをWebアプリで提供する際の注意点

医療分野では特に以下が重要：

- **規制**: 医療目的のソフトは医療機器プログラム（SaMD）に該当し得るため、該当性判断と要求事項の整理が必要
- **プライバシー**: 患者画像の取り扱い、同意、保管、アクセス制御、ログ管理などを厳格に
- **セキュリティ**: 暗号化、脆弱性対応、権限管理、サイバー対策
- **表示と運用**: 意図された使用、適用範囲、限界、想定外使用への警告、過信防止の設計

### 3. 医療現場でのAI導入の課題

導入・運用で問題になりやすい点：

- 医療従事者の受容と教育（使い方・限界の理解）
- 既存ワークフロー（PACS/EHR等）との統合
- 責任分界とインシデント対応
- 導入・運用・保守コスト
- 継続的な性能監視（データドリフト、施設差、時間変化）と変更管理

### 設問演習の回答例

### 設問1：公開AIのリスク分析

**(a)** 各観点からのリスク：

**技術的観点：**
- スマートフォン撮影画像とMedMNISTの28×28ピクセル画像では、解像度・撮影条件・照明・角度が大きく異なり、モデルの性能が著しく低下する可能性がある
- モデルは特定のデータセット（MedMNIST）で学習しており、実際の多様な皮膚病変に対する汎化性能は検証されていない
- 入力画像の品質（ブレ、影、部位の写り方）によって予測結果が大きく変動し得る

**法規制の観点：**
- 診断目的で使用する場合、医療機器プログラム（SaMD: Software as a Medical Device）に該当し得る。日本では薬機法に基づく承認・認証が必要となる場合がある
- 誤診による健康被害が生じた場合の責任の所在が不明確
- 医師法との関係で、AIによる診断行為の位置づけを整理する必要がある

**倫理的観点：**
- アップロードされた患者画像のプライバシー保護（保存・第三者アクセス・ログ管理）が不十分
- 患者がAIの結果を過信し、医療機関の受診を遅らせるリスク
- AIの結果が「正常」であっても実際には異常がある場合、患者に誤った安心を与える可能性

**(b)** 免責事項・注意書きの例：

「本アプリは教育・研究目的のデモであり、医療診断には使用できません。結果は参考情報であり、正確性を保証するものではありません。皮膚の異常が気になる場合は、必ず医療機関を受診してください。」

### 設問2：AI導入の全体設計

**(a)** そのまま導入することは**適切でない**。最低限クリアすべき条件：

1. **データの適合性**: 導入先の病院の病理画像（実解像度・染色条件・スキャナ）で評価し直す必要がある。28×28ピクセルの教材データでの性能は、実臨床の性能を反映しない
2. **外部検証**: 開発に使ったデータとは独立したデータ（導入先の施設のデータや、他施設データ）で性能を検証する。内部検証のみでは汎化性能を保証できない
3. **規制・承認対応**: 診断支援として使用する場合、医療機器プログラムとしての該当性を確認し、必要に応じて承認プロセスを経る。使用目的・対象・限界を明確に文書化する

**(b)** 性能低下の原因と検知方法：

**考えられる原因：**
- **データドリフト**: 染色試薬のロット変更、スキャナの更新・劣化、撮影プロトコルの変更などにより、入力画像の分布が学習時と徐々にずれる
- **コンセプトドリフト**: 診断基準の改訂、新しい疾患分類の導入などにより、ラベルの定義自体が変化する
- **患者集団の変化**: 病院の診療圏や紹介パターンの変化により、症例の構成が変わる

**検知方法：**
- 定期的なサンプリング評価：一定期間ごとに、医師の診断結果とAIの予測を照合し、性能指標（感度・特異度等）の推移を監視する
- 入力データの分布監視：画像の統計的特徴量（明度分布、コントラスト等）を継続的に記録し、学習時のデータとの乖離を検出する
- インシデント報告の収集：臨床現場からのフィードバック（明らかな誤判定の報告）を体系的に収集する仕組みを設ける
